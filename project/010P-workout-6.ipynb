```python
import pandas as pd
```

```python
df = pd.read_csv('./data/nyc_taxi_2019-01.csv', usecols=['passenger_count', 'trip_distance', 'total_amount'])
```

```python
df.sort_values(by='trip_distance', ascending=False).head(20).mean().round(2)
```

```python
df.sort_values('trip_distance', ascending=False)['total_amount'].iloc[:20].mean().round(2)
```

```python
df.sort_values(by='trip_distance', ascending=False)['total_amount'].head(20).mean().round(2)
```

```python
df.sort_values('trip_distance')['total_amount'].iloc[-20:].mean().round(2)
```

```python
df.sort_values(by='trip_distance')['total_amount'].tail(20).mean().round(2)
```

```python
(df.sort_values(['passenger_count', 'trip_distance'], ascending=[True, False])['total_amount'].head(50).mean().round(2))
```

```python
(df.sort_values(['passenger_count', 'trip_distance'], ascending=[True, False])['total_amount'].iloc[:50].mean().round(2))
```
