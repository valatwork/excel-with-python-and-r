Load the CSV file `nyc_taxi_2019-01.csv` into a data frame using only the columns `passenger_count`, `trip_distance`, and `total_amount`

```python
import pandas as pd
```

```python
df = pd.read_csv('./data/nyc_taxi_2019-01.csv', usecols=['passenger_count', 'trip_distance', 'total_amount'])
```

Using a *descending* sort, find the average cost of the 20 longest (in distance) taxi
rides in January 2019.

```python
df.sort_values(by='trip_distance', ascending=False).head(20).mean().round(2)
```

```python
df.sort_values('trip_distance', ascending=False)['total_amount'].iloc[:20].mean().round(2)
```

```python
df.sort_values(by='trip_distance', ascending=False)['total_amount'].head(20).mean().round(2)
```

Using an *ascending* sort, find the average cost of the 20 longest (in distance) taxi
rides in January 2019.

```python
df.sort_values('trip_distance')['total_amount'].iloc[-20:].mean().round(2)
```

```python
df.sort_values(by='trip_distance')['total_amount'].tail(20).mean().round(2)
```

Sort by ascending passenger count and descending trip distance. What is the average price paid for the top 50 rides?

```python
(df.sort_values(['passenger_count', 'trip_distance'], ascending=[True, False])['total_amount'].head(50).mean().round(2))
```

```python
(df.sort_values(['passenger_count', 'trip_distance'], ascending=[True, False])['total_amount'].iloc[:50].mean().round(2))
```

In which five rides did people pay the most per mile? How far did people go on
those trips?

```python
df = df[df['trip_distance'] != 0]
df['cost_per_mile'] = df['total_amount'] / df['trip_distance']
df['cost_per_mile'] = df['cost_per_mile'].round(2)
```

```python
df.sort_values('cost_per_mile').tail(5).round(2)
```

Let’s assume that multipassenger rides are split evenly among the passengers.
Given that assumption, in which 10 multipassenger rides did each individual
pay the greatest amount?

```python
df = df[df['passenger_count'] >= 2]
df['payment_per_person'] = df['total_amount'] / df['passenger_count']
df.sort_values('payment_per_person').tail(10)
```

Using `ignore_index=True`, find the mean `total_amount` of the 20 longest rides:

```python
df.sort_values('trip_distance',
               ascending=False,
                ignore_index=True)['total_amount'].loc[:20].mean().round(2)
```

# 30

Load taxi data from January 2019 into a data frame using only the columns
`passenger_count`, `trip_distance`, and `total_amount`.

```python
df = pd.read_csv('./data/nyc_taxi_2019-01.csv', usecols=['passenger_count', 'trip_distance', 'total_amount'])
```

For each number of passengers, find the mean cost of a taxi ride. Sort this
result from lowest (i.e., cheapest) to highest (i.e., most expensive).

```python
df.groupby('passenger_count')['total_amount'].mean().round(2).sort_values()
```

Sort the results again by increasing the number of passengers.

```python
df.groupby('passenger_count')['total_amount'].mean().round(2).sort_index()
```

Create a new column, `trip_distance_group`, in which the values are short (< 2
miles), medium ($\geq$ 2 miles and $\leq$ 10 miles), and long (> 10 miles). What is the
average number of passengers per trip length category? Sort this result from highest (most passengers) to lowest (fewest passengers).

```python
df['trip_distance_group'] = pd.cut(df['trip_distance'],
                                   [df['trip_distance'].min(), 2, 10,
                                    df['trip_distance'].max()],
                                  labels=['short', 'medium', 'long'])
df.groupby('trip_distance_group', observed=False)['passenger_count'].mean().round(2).sort_values(ascending=False)
```

Create a single data frame containing rides from both January 2019 and January 2020, with a column year indicating which year the ride comes from. Use
`groupby`  to compare the average cost of a taxi in January from each of these two
years.

```python
jan_2019_filename = './data/nyc_taxi_2019-01.csv'
jan_2020_filename = './data/nyc_taxi_2020-01.csv'
```

```python
jan_2019_df = pd.read_csv(jan_2019_filename,
                usecols=['passenger_count',
                         'trip_distance', 'total_amount'])
jan_2019_df['year'] = 2019
jan_2020_df = pd.read_csv(jan_2020_filename,
                usecols=['passenger_count',
                         'trip_distance', 'total_amount'])
jan_2020_df['year'] = 2020
```

```python
df = pd.concat([jan_2019_df, jan_2020_df])
```

```python
df.groupby('year')['total_amount'].mean().round(2)
```

Create a two-level grouping, first by year and then by `passenger_count`.

```python
df.groupby(['year', 'passenger_count'])['total_amount'].mean().round(2)
```

Use `corr` and then sort_values to find which columns have the highest correlation.

```python
df.corr().sort_values('passenger_count').round(4)
```

# 31

Load the `oecd_tourism.csv` file into a data frame. Use only the columns `LOCATION`, `SUBJECT`, `TIME`, and `Value`.

```python
tourism_filename = './data/oecd_tourism.csv'
tourism_df = pd.read_csv(tourism_filename,
                        usecols=['LOCATION', 'SUBJECT', 'TIME', 'Value'])

tourism_df.head()
```

Find the five countries that received the greatest amount of tourist dollars, on
average, across years in the data set.

```python
(
    tourism_df
    .loc[tourism_df['SUBJECT'] == 'INT_REC']
    .groupby('LOCATION')['Value']
    .mean()
    .sort_values(ascending=False)
    .head()
    .round(2)
)
```

Find the five countries whose citizens spent the least amount of tourist dollars,
on average, across years in the data set.

```python
(
    tourism_df
    .loc[tourism_df['SUBJECT'] == 'INT-EXP']
    .groupby('LOCATION')['Value']
    .mean()
    .sort_values()
    .head()
    .round(2)
)
```

```python
locations_filename = './data/oecd_locations.csv'
locations_df = pd.read_csv(locations_filename,
                           header=None,
                           names=['LOCATION', 'NAME'],
                           index_col='LOCATION')
```

Load the `oecd_locations.csv` file into a data frame. Use the `LOCATION` column as the index

```python
fullname_df = locations_df.join(tourism_df.set_index('LOCATION'))
fullname_df.head()
```

Join the two data frames, `tourism_df` and `locations_df`, using the `LOCATION` column as the index. What is the name of the country with the most tourist dollars received?

```python
(
    fullname_df
    .loc[fullname_df['SUBJECT'] == 'INT_REC']
    .groupby('NAME')['Value']
    .mean()
    .sort_values(ascending=False)
    .head()
    .round(2)
)
```

What is the name of the country with the least amount of tourist dollars spent?

```python
(
    fullname_df
    .loc[fullname_df['SUBJECT'] == 'INT-EXP']
    .groupby('NAME')['Value']
    .mean()
    .sort_values()
    .head()
    .round(2)
)
```

What happens if you perform the join in the other direction? That is, what if
you invoke join on `tourism_df`, passing it an argument of `locations_df`? Do
you get the same result?

```python
emanlluf_df = tourism_df.set_index('LOCATION').join(locations_df)
emanlluf_df.info()
```

Get the mean tourism income per year rather than by country. Do you see any
evidence of less tourism income during the time of the Great Recession, which
started in 2008?

```python
(
    fullname_df
    .loc[fullname_df['SUBJECT'] == 'INT_REC']
    .groupby('TIME')['Value']
    .mean()
    .sort_values(ascending=False)
    .round(2)
)
```

Reset the index on `locations_df` such that it has a (default) numeric index and
two columns (`LOCATION` and `NAME`). Now run join on `locations_df`, specifying
that you want to use the `LOCATION` column on the caller rather than its index. (The
data frame passed as an argument to join will always be joined on its index.)

```python
tourism_df = tourism_df.set_index('LOCATION')
locations_df.join(tourism_df, on='LOCATION')
```
