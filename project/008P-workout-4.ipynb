1 Create a data frame from the file nyc-parking-violations-2020.csv. We are only interested in a handful of the columns:
- Date First Observed
- Plate ID
- Registration State
- Issue Date (a string in MM/DD/YYYY format, always followed by 12:00:00
- Vehicle Make
- Street Name
- Vehicle Color

```python
nyc_pv = pd.read_csv('./data/nyc-parking-violations-2020.csv',
                    usecols=['Date First Observed','Plate ID', 'Registration State', 'Issue Date', 'Vehicle Make', 'Street Name', 'Vehicle Color'])
```

Set the data frame’s index to the Issue Date column.

```python
df = nyc_pv.set_index('Issue Date')
```

Determine what three makes were most frequently ticketed on January 2, 2020

```python
df.loc['01/02/2020 12:00:00 AM', 'Vehicle Make'].value_counts().head(3)
```

Determine the five streets on which cars got the most tickets on June 1, 2020.

```python
df.loc['06/01/2020 12:00:00 AM', 'Street Name'].value_counts().head(5)
```

Set the index to `Vehicle Color`.

```python
df = df.reset_index().set_index('Vehicle Color')
```

Determine the most common make of vehicles that were either red or blue

```python
(df.loc[['BLUE', 'RED'], 'Vehicle Make'].value_counts().head(1))
```

What three car makes were most often ticketed from January 2 through January 10?

```python
df = df.reset_index()   # undo the setting of the index from the previous cell

# let's do it in chained format:

(
    df
    .set_index('Issue Date')
    .sort_index()
    .loc['01/02/2020 12:00:00 AM':'01/10/2020 23:59:59 PM',
         'Vehicle Make']
    .value_counts()
    .head(3)
)
```

How many tickets did the second-most-ticketed car get in 2020?

```python
df = df.reset_index()
df['Plate ID'].value_counts().head(2)
```

What state was that car from?

```python
df = df.set_index('Plate ID')
df.loc['2704819', 'Registration State']
```

Was it always ticketed in the same location?

```python
df.loc['2704819', 'Street Name'].value_counts()
```

Read in the scores file, only keeping the `Year`, `State.Code`, `Total.Math`, `Total.Test-takers`, and `Total.Verbal` columns.

```python
sat_scores = pd.read_csv('./data/sat-scores.csv',
                         usecols= ['Year', 'State.Code', 'Total.Math', 'Total.Test-takers', 'Total.Verbal'])
```

Create a multi-index based on the year and the two-letter state code.

```python
df = sat_scores.set_index(['Year', 'State.Code'])
```

```python
df.info()
```

Determine how many people took the SAT in 2005.

```python
df.loc[2005, 'Total.Test-takers'].sum()
```

Determine the average SAT math score in 2010 from New York (NY), New Jersey (NJ), Massachusetts (MA), and Illinois (IL).

```python
df.loc[(2010, ['NY', 'NJ', 'MA', 'IL']), 'Total.Math'].mean()
```

Determine the average SAT verbal score in 2012–2015 from Arizona (AZ), California (CA), and Texas (TX).

```python
df.loc[([2012,2013,2014,2015], ['AZ', 'CA', 'TX']), 'Total.Verbal'].mean()
```

Read the data file (`olympic_athlete_events.csv`) into a data frame, keep these columns: `Age`, `Height`, `Team`, `Year`, `Season`, `City`, `Sport`,
`Event`, and `Medal`.

The multi-index should be based on `Year`, `Season`, `Sport`, and `Event`.

```python
olympic_athlete_events = pd.read_csv('./data/olympic_athlete_events.csv',
                                     usecols=['Age', 'Height', 'Team', 'Year', 'Season', 'City', 'Sport', 'Event', 'Medal'],
                                     index_col= ['Year', 'Season', 'Sport', 'Event'])
```

```python
olympic_athlete_events.info()
```

```python
df = olympic_athlete_events.sort_index()
df.head()
```

What is the average age of the competing athletes in summer games held between 1936 and 2000?

```python
df.loc[(slice(1936,2000), 'Summer'), 'Age'].mean()
```

Which team won the most medals in all archery events?

```python
df.dropna(subset='Medal').loc[(slice(None), 'Summer', 'Archery'), 'Team'].value_counts()
```

Starting in 1980, what is the average height of the “Table Tennis Women’s Team” event?

```python
df.loc[(slice(1980, None), 'Summer', slice(None), 'Table Tennis Women\'s Team'), 'Height'].mean()
```

Starting in 1980, what is the average height of both “Table Tennis Women’s Team” and “Table Tennis Men’s Team”?

```python
df.loc[(slice(1980, None), 'Summer', slice(None), ['Table Tennis Men\'s Team', 'Table Tennis Women\'s Team']), 'Height'].mean()
```

Using `xs`, we can shorten that query to:

```python
df.xs('Table Tennis Women\'s Team', level='Event')['Height'].mean()
# if we didn't know the season, we could use:
# df.xs(('Summer', 'Table Tennis Women\'s Team'), level=['Season', 'Event'])['Height'].mean()
```

A more general way to retrieve from a multi-index is `IndexSlice`.

```python
from pandas import IndexSlice as idx
```

```python
df.loc[idx[1980:2016, :, 'Swimming':'Table tennis'], :]
```

How tall was the tallest-ever tennis player in Olympic games from 1980 until 2016?

Since Tennis is a summer sport, let's remove `Season` from the index.

```python
df = df.reset_index('Season')
```

```python
df.loc[(slice(1980,2020), 'Tennis'), 'Height'].max()
```

In which city were the most gold medals awarded from 1980 onward?

```python
df.loc[1980:].loc[lambda df_: df_['Medal'] == 'Gold', 'City'].value_counts()
```

How many gold medals were received by the United States since 1980?

```python
df.loc[1980:].loc[lambda df_: (df_['Team'] == 'United States') & (df_['Medal'] == 'Gold'), 'City'].count()
```
