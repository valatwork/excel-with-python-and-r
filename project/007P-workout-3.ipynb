Create a data frame from two different CSV files containing New York taxi data: one from July 2019 (before the pandemic) and a second from July 2020 (near the height of the pandemic, at least in New York).
The dataframe should contain three columns from the files: `passenger_count`, `total_amount`, and `payment_type`.
It should also include a fifth column, `year`, which should be set to 2019 or 2020, depending on the file from which the data was loaded.

```python
df_2019_jul = pd.read_csv('./data/nyc_taxi_2019-07.csv', usecols=['passenger_count', 'total_amount', 'payment_type'])
df_2019_jul['year'] = 2019
df_2020_jul = pd.read_csv('./data/nyc_taxi_2020-07.csv', usecols=['passenger_count', 'total_amount', 'payment_type'])
df_2020_jul['year'] = 2020
df = pd.concat([df_2019_jul, df_2020_jul])
```

How many rides were taken in 2019 and 2020, and what is the difference between these two figures?

```python
(df.loc[df['year'] == 2019, 'total_amount'].count() - df.loc[df['year'] == 2020, 'total_amount'].count())
```

How much money (in total) was collected in 2019 and 2020, and what was the difference between these two figures, rounded to 2 decimal places?

```python
df.loc[df['year'] == 2019, 'total_amount'].sum().round(2) - df.loc[df['year'] == 2020, 'total_amount'].sum().round(2)
```

 Did the proportion of trips with more than one passenger change dramatically?

```python
df.loc[(df['year'] == 2019) & (df['passenger_count'] > 1), 'passenger_count'].count() / df.loc[df['year'] == 2019, 'payment_type'].count()
```

```python
df.loc[(df['year'] == 2020) & (df['passenger_count'] > 1), 'passenger_count'].count() / df.loc[df['year'] == 2020, 'payment_type'].count()
```

Did people use cash (i.e., `payment_type` of 2) less in 2020 than in 2019?

```python
df.loc[(df['year'] == 2019) & (df['payment_type'] == 2), 'payment_type'].count() / df.loc[df['year'] == 2019, 'payment_type'].count()
```

```python
df.loc[(df['year'] == 2020) & (df['payment_type'] == 2), 'payment_type'].count() / df.loc[df['year'] == 2020, 'payment_type'].count()
```

Use the corr method on df to find the correlations among the columns

Because there are some missing values in the data, the default behavior of pandas is to use the `float64` type for all columns. This is not necessary, as we can use smaller types for each column.

Before we can do that, however, the missing values must be removed. Let's start by re-importing the data and change the `dtype` of each column to the smallest possible type *before* removing the missing values:

```python
df = pd.read_csv(
    './data/nyc_taxi_2020-07.csv',
    usecols=['passenger_count', 'total_amount', 'payment_type'],
    dtype={
        'passenger_count': 'float32',
        'total_amount': 'float32',
        'payment_type': 'float32',
    },
)
```

```python
df.info()
```

Now we can remove the missing values:

```python
df = df.dropna().copy()
```

And finally, we can convert the columns to the smallest possible type:

```python
df['passenger_count'] = df['passenger_count'].astype('int8')
df['payment_type'] = df['payment_type'].astype('int8')
```

Retrieve 1 year of [historical S&P 500 data from Yahoo Finance](https://finance.yahoo.com/quote/%5EGSPC/history?p=%5EGSPC).

Note that Yahoo seems to look at the `User-Agent` header in the HTTP request, which cannot be set in `read_html`. So you'll need to use `requests` to retrieve the data, setting `User-Agent` to a string equal to `'Mozilla 5.0'`. Turn the content of the result into a `StringIO`, and then finally feed that to `read_html` and retrieve the data.

```python
import requests
from io import StringIO

r = requests.get('https://finance.yahoo.com/quote/%5EGSPC/history?p=%5EGSPC',
                headers={'User-Agent': 'Mozilla/5.0'})

df = pd.read_html(StringIO(r.content.decode()))[0].set_index('Date').iloc[:-1]
```

```python
df.info()
```

Rename the columns to be more user-friendly. The first column should be renamed to `Date`, and the last two columns should be renamed to `Close` and `Adj Close`, respectively.

To convert the `growth_from_2000_to_2013` column into a floating-point number, first remove the trailing `%` sign:

```python
# Remove trailing %
df['growth_from_2000_to_2013'] = df['growth_from_2000_to_2013'].str.rstrip('%')
```

Find the empty strings, and set their value to `0`:

```python
# Find empty strings, and turn to 0
df.loc[df['growth_from_2000_to_2013'] == '', 'growth_from_2000_to_2013'] = '0'
```

Then find the mean and median changes in city size between 2000 and 2013.

```python
# Turn to float, return mean + median
df['growth_from_2000_to_2013'].astype(float).describe()[['mean', '50%']]
```

How many cities had positive growth in this period, and how many had negative growth?

```python
# First ensure it's numeric
df['growth_from_2000_to_2013'] = df['growth_from_2000_to_2013'].astype(float)

# Check min value and create appropriate bins
min_val = df['growth_from_2000_to_2013'].min()

if min_val < 0:
    # If we have negative values, create bins for negative and positive
    bins = [min_val, 0, df['growth_from_2000_to_2013'].max()]
    result = pd.cut(df['growth_from_2000_to_2013'],
                  bins=bins,
                  include_lowest=True,
                  labels=['-', '+']).value_counts()
else:
    # If all values are positive, just count them directly
    result = pd.Series([0, len(df)], index=['-', '+'])

result
```
