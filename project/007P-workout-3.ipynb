```python
import pandas as pd
```

```python
df = pd.read_csv('project/data/nyc_taxi_2019-01.csv', usecols=['passenger_count', 'trip_distance', 'total_amount', 'payment_type'])
```

How many taxi rides had more than eight passengers?

```python
df.loc[df['passenger_count'] > 8, 'passenger_count'].count()
```

How many taxi rides had zero passengers?


```python
df.loc[df['passenger_count'] == 0, 'passenger_count'].count()
```

How many taxi rides were paid for in cash and cost over $1,000?

```python
df.loc[(df['payment_type'] == 2) & (df['total_amount'] > 1000), 'passenger_count'].count()
```

How many rides cost less than $0?

```python
df.loc[df['total_amount'] < 0].shape[0]
```

How many rides traveled a below-average distance and cost an above-average amount?

```python
df.loc[(df['trip_distance'] < df['trip_distance'].mean()) & (df['total_amount'] > df['total_amount'].mean()), 'trip_distance'].count()
```

Repeat this exercise using the query method rather than a boolean index and loc.

```python
df.query('passenger_count > 8').shape[0]
```

```python
df.query('passenger_count == 0').shape[0]
```

```python
df.query('payment_type == 2 and total_amount > 1000').shape[0]
```

```python
df.query('total_amount < 0').shape[0]
```

```python
df.query('trip_distance < @df["trip_distance"].mean() and total_amount > @df["total_amount"].mean()').shape[0]
```

How many rides that cost less than $0 involved either a dispute (`payment_type` of 4) or a voided trip (`payment_type` of 6)?

```python
df.loc[(df['total_amount'] < 0) & (df['payment_type'].isin([4, 6]))].shape[0]
```

```python
df.query('total_amount < 0 and payment_type in [4, 6]').shape[0]
```
